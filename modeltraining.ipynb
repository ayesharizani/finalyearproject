```python
X_train = data0.drop('label', axis=1)
y_train = data0['label']

X_test = data1.drop('label', axis=1)
y_test = data1['label']

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

model training

```python
import keras
import tensorflow as tf
from sklearn import metrics
from keras import regularizers
from keras.models import Model
from keras.layers import Conv1D, MaxPooling1D, Flatten, Dropout, Dense, Input
from sklearn.metrics import confusion_matrix, classification_report
```

```python
models = []
accuracy_train = []
accuracy_test = []

def storeResults(model, a, b):
  models.append(model)
  accuracy_train.append(round(a, 3))
  accuracy_test.append(round(b, 3))
```

Autoencoder

```python
input_dim = X_train.shape[1]
encoding_dim = input_dim

input_layer = Input(shape=(input_dim, ))

encoder = Dense(encoding_dim, activation="relu",
                activity_regularizer=regularizers.l1(10e-4))(input_layer)
encoder = Dense(encoding_dim, activation="relu")(encoder)
encoder = Dense(encoding_dim - 2, activation="relu")(encoder)

code = Dense(encoding_dim - 4, activation='relu')(encoder)

decoder = Dense(encoding_dim - 2, activation='relu')(code)
decoder = Dense(encoding_dim, activation='relu')(decoder)
output_layer = Dense(input_dim, activation='relu')(decoder)

autoencoder = Model(inputs=input_layer, outputs=output_layer)
autoencoder.summary()
```

```python
autoencoder.compile(optimizer='adam',
                    loss='binary_crossentropy',
                    metrics=['accuracy'])

history = autoencoder.fit(X_train, X_train, epochs=20, batch_size=64, shuffle=True, validation_split=0.2, class_weight={0:1, 1:1})
```

```python
acc_train_auto = autoencoder.evaluate(X_train, X_train)[1]
acc_test_auto = autoencoder.evaluate(X_test, X_test)[1]

print('\nAutoencoder: Accuracy on training data: {:.3f}' .format(acc_train_auto))
print('Autoencoder: Accuracy on testing data: {:.3f}' .format(acc_test_auto))
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Autoencoder Model Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Autoencoder Model Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
storeResults('Autoencoder', acc_train_auto, acc_test_auto)
```

Multilayer Perceptron

```python
input_dim = X_train.shape[1]
input_layer = Input(shape=(input_dim, ))

dense = Dense(64, activation='relu')(input_layer)
dense = Dense(32, activation='relu')(dense)
dense = Dropout(0.3)(dense)
dense = Dense(16, activation='relu')(dense)
output = Dense(1, activation='sigmoid')(dense)

mlp_model = Model(inputs=input_layer, outputs=output)
mlp_model.summary()
```

```python
mlp_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

mlp_history = mlp_model.fit(X_train, y_train, epochs=20, batch_size=64, validation_split=0.2, shuffle=True, class_weight={0:1, 1:1})
```

```python
acc_train_mlp = mlp_model.evaluate(X_train, y_train)[1]
acc_test_mlp = mlp_model.evaluate(X_test, y_test)[1]

print('\nMLP: Accuracy on training data: {:.3f}'.format(acc_train_mlp))
print('MLP: Accuracy on testing data: {:.3f}'.format(acc_test_mlp))
```

```python
final_train_acc = mlp_history.history['accuracy'][-1]
final_val_acc = mlp_history.history['val_accuracy'][-1]
final_train_loss = mlp_history.history['loss'][-1]
final_val_loss = mlp_history.history['val_loss'][-1]

print(f"Final Training Accuracy: {final_train_acc:.4f}")
print(f"Final Validation Accuracy: {final_val_acc:.4f}")
print(f"Final Training Loss: {final_train_loss:.4f}")
print(f"Final Validation Loss: {final_val_loss:.4f}")
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(mlp_history.history['accuracy'], label='Train Accuracy')
plt.plot(mlp_history.history['val_accuracy'], label='Validation Accuracy')
plt.title('MLP Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(mlp_history.history['loss'], label='Train Loss')
plt.plot(mlp_history.history['val_loss'], label='Validation Loss')
plt.title('MLP Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
y_pred_mlp = mlp_model.predict(X_test)
y_pred_mlp_labels = (y_pred_mlp > 0.5).astype("int32")

print("Confusion Matrix - MLP Model:")
print(confusion_matrix(y_test, y_pred_mlp_labels))

print("\nClassification Report - MLP Model:")
print(classification_report(y_test, y_pred_mlp_labels))
```

```python
storeResults('Multilayer Perceptron', acc_train_mlp, acc_test_mlp)
```

Convolutional Neural Network

```python
X_train_cnn = np.expand_dims(X_train, axis=2)
X_test_cnn = np.expand_dims(X_test, axis=2)

input_dim = X_train_cnn.shape[1]
input_layer = Input(shape=(input_dim, 1))

conv = Conv1D(filters=64, kernel_size=3, activation='relu')(input_layer)
pool = MaxPooling1D(pool_size=2)(conv)
flat = Flatten()(pool)
dense = Dense(32, activation='relu')(flat)
drop = Dropout(0.3)(dense)
output = Dense(1, activation='sigmoid')(drop)

cnn_model = Model(inputs=input_layer, outputs=output)
cnn_model.summary()
```

```python
cnn_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

cnn_history = cnn_model.fit(X_train_cnn, y_train, epochs=20, batch_size=64, validation_split=0.2, shuffle=True, class_weight={0:1, 1:1})
```

```python
acc_train_cnn = cnn_model.evaluate(X_train_cnn, y_train)[1]
acc_test_cnn = cnn_model.evaluate(X_test_cnn, y_test)[1]

print('\nCNN: Accuracy on training Data: {:.3f}'.format(acc_train_cnn))
print('CNN: Accuracy on testing Data: {:.3f}'.format(acc_test_cnn))
```

```python
final_train_acc = cnn_history.history['accuracy'][-1]
final_val_acc = cnn_history.history['val_accuracy'][-1]
final_train_loss = cnn_history.history['loss'][-1]
final_val_loss = cnn_history.history['val_loss'][-1]

print(f"Final Training Accuracy: {final_train_acc:.4f}")
print(f"Final Validation Accuracy: {final_val_acc:.4f}")
print(f"Final Training Loss: {final_train_loss:.4f}")
print(f"Final Validation Loss: {final_val_loss:.4f}")
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(cnn_history.history['accuracy'], label='Train Accuracy')
plt.plot(cnn_history.history['val_accuracy'], label='Validation Accuracy')
plt.title('CNN Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(cnn_history.history['loss'], label='Train Loss')
plt.plot(cnn_history.history['val_loss'], label='Validation Loss')
plt.title('CNN Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
y_pred_cnn = cnn_model.predict(X_test_cnn)
y_pred_cnn_labels = (y_pred_cnn > 0.5).astype("int32")

print("Confusion Matrix - CNN Model:")
print(confusion_matrix(y_test, y_pred_cnn_labels))

print("\nClassification Report - CNN Model:")
print(classification_report(y_test, y_pred_cnn_labels))
```

```python
storeResults('Convolutional Neural Network', acc_train_cnn, acc_test_cnn)
```
